#Creating tensors


**Creation (The Builders):**
torch.tensor(): Creates a tensor from data (like a Python list).

torch.zeros() / torch.ones(): Generates tensors filled with 0s or 1s (useful for initializing weights or masks).

torch.rand(): Creates a tensor with random numbers (essential for starting training).







**Reshaping (The Organizers):**
tensor.view() / torch.reshape(): Changes the dimensions of your data without changing the data itself (e.g., flattening an image).

torch.cat(): Joins two tensors together.

tensor.permute(): Rearranges dimensions (e.g., swapping rows and columns).







**Math & Compute (The Calculators):**
torch.matmul(): Performs matrix multiplication, the engine behind nearly all neural networks.

tensor.to(): Moves your tensor to the GPU (e.g., .to('cuda')) for faster processing.

tensor.item(): Extracts a single standard Python number from a one-element tensor.
