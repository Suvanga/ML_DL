```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.datasets import fetch_california_housing
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df =fetch_california_housing()
```

```python
df
```

```python
dataset = pd.DataFrame(df.data)
dataset.columns = df.feature_names
dataset.head()
```

```python
# df.data and price is the dependent featurew which we are going to put it in the dataframe
```

```python
dataset["price"] =df.target
```

```python
dataset.head()
```

#Spliiting dependent and independent feature


```python
X = dataset.iloc[:,:-1]
y = dataset.iloc[:,-1]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.33, random_state=42)

lin_reg = LinearRegression()
lin_reg.fit(X_train,y_train)
```

##Dividing the train and test data

```python
MSE = cross_val_score(lin_reg,X_train,y_train, scoring = 'neg_mean_squared_error', cv =5)
```

```python
MSE
```

```python
# 1. Train the model first
lin_reg.fit(X, y)

# 2. Now you can predict
lin_reg.predict(X)
```

##RIDGE REGRESSION

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
```

```python
ridge = Ridge()
params = {'alpha': [1e-15, 1e-10, 1e-10, 1e-8,1e-3,1e-2,1,5,10,20]}
ridge_regressor = GridSearchCV(ridge, params, scoring = 'neg_mean_squared_error', cv = 5)
ridge_regressor.fit(X_train,y_train)
```

```python
print(ridge_regressor.best_params_)
print(ridge_regressor.best_score_)
```

##lasso Regression

```python
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV
lasso = Lasso()
params = {'alpha': [1e-15, 1e-10, 1e-10, 1e-8,1e-3,1e-2,1,5,10,20]}
Lasso_regressor  = GridSearchCV(lasso, params, scoring = 'neg_mean_squared_error', cv = 5)
Lasso_regressor.fit(X_train,y_train)
```

```python
print(Lasso_regressor.best_params_)
print(Lasso_regressor.best_score_)
```

##Printing the R square

```python
from sklearn.metrics import r2_score

y_pred = lin_reg.predict(X_test)
Y_Ridge_predict = ridge_regressor.predict(X_test)
Y_Lasso_predict = Lasso_regressor.predict(X_test)
```

```python
Lin_r2score = r2_score(y_pred, y_test)
Ridge_r2score = r2_score(Y_Ridge_predict, y_test)
Lasso_r2score = r2_score(Y_Lasso_predict, y_test)

#Adjusted r2 are
```

```python
print(Lin_r2score)
print(Ridge_r2score)
print(Lasso_r2score)
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

df = load_breast_cancer()
```

```python
X = pd.DataFrame(df['data'], columns = df['feature_names'])
```

```python
##Dependent featues are
y = pd.DataFrame(df['target'], columns = ['Target'])
```

```python
y
```

```python
## we have to check if the data set is balanced or not balanced

y['Target'].value_counts()
```

```python
##Train Test Split
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.33, random_state=42)
```

```python
params = [{'C':[1,5,10]},{'max_iter':[100,150]}]
model1 =LogisticRegression(C=100, max_iter =100)
```

```python
model = GridSearchCV(model1, param_grid = params, scoring = 'f1', cv= 5)
```

```python
model.fit(X_train, y_train)
```

```python
model.best_score_
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
print(confusion_matrix(y_test, y_pred))
```

```python
accuracy_score(y_test,y_pred)
```

```python
classification_report(y_test, y_pred)
```
